1. Explain how implicit missing values impact medical datasets, and justify your choice of imputation strategy when handling these zeros compared to outright row deletion.


*   Implicit missing values can cause problems because models may treat missing measurements as real values. Thinks like zeros in glucose or blood pressure can represent missing information. Median imputation was used because it replaces these values with reasonable estimates and is less affected by extreme values. This is better than deleting rows because removing patients with missing measurements would reduce the amount of data available for training and could introduce bias.


2. Contrast how hard voting aggregates predictions versus soft voting. Why do well-calibrated probability estimates matter when implementing soft voting across heterogeneous models?


*   Hard voting uses majority rule meaning the class predicted by the most models becomes the final prediction. Soft voting averages the probabilities from each model before making a decision. Soft voting can see how confident each model is rather than just its final prediction. Well calibrated probabilities are important because an model thats too confident could influence the final result too much even when its predictions are incorrect.


3. What is the role of out-of-bag evaluation in Bagging classifiers, and why does an OOB score serve as an efficient alternative to traditional cross-validation splits during ensemble training?


*   Out-of-bag evaluation measures a Bagging model's performance using training observations that were not used for individual trees during  sampling. These unused observations help estimate how well the model performs on unseen data. OOB evaluation is very efficient because it provides an validation estimate without repeatedly training the entire ensemble on different splits as what cross-validation does. Comparing OOB accuracy with test accuracy helps determine whether the model performs better.


4. How does introducing random feature subsetting in Random Forests differ from standard Bagging, and how does this feature diversity impact model variance and generalization?


*   Standard Bagging trains multiple Decision Trees using different random samples of the training data but each tree can consider all features when splitting. Random Forest adds another level of randomness to it by selecting only a subset of features at each split. This creates more variables in the trees and reduces the chance that they all make similar mistakes. So random Forest can reduce model variance and improve predictions on unseen data.
