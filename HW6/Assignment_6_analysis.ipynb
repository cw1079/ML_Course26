**3. Why is it critical to fit the PCA transformer strictly on the training split rather than on the entire dataset prior to splitting? What negative impacts occur if PCA is fitted on the full dataset before training downstream classifiers?**

*   The PCA should only be fitted on the training data to prevent data leakage. If PCA is fitted on the entire dataset before splitting, it learns patterns from both the training and testing images. This means the test data has already made an influence to the model about how the images are processed before they can be evaluated. The accuracy could appear higher than it really is which can make the results less reliable for this problem. Fitting the PCA only on the training data keeps the testing images separate and gives a more fair measurement of how well the model will do on new images.
